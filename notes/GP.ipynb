## 随机变量
在某个时刻的量，做足够多的realization会呈现所选取的分布，这里是高斯分布。

```python
import numpy as np

# RNG = RANDOM NUMBER GENERATOR, 1 is seed. Here, we create a random number creator with numpy
rng = np.random.default_rng(1)

# with the generator rng, we input parameters loc(here is mean value), scale(here the width), size(here the one dim paramiter, 10_000, _ to make it readable)
x_at_dat10 = rng.normal(loc=0.0, scale=1.0, size = 10_000)
```
